## Data Science: Worum geht es überhaupt?

*Modul 1 · Freitag 16.10*

Rollen, CRISP-DM, ML-Typen

**Data Science macht Daten nützlich.** Aus Zahlen werden Entscheidungen. Dafür braucht es vier Dinge zusammen: Mathematik und Statistik, Programmieren, Kommunikation und Fachwissen über das Geschäft.

*[Diagramm: Fehlt eine Säule, wird es schwierig: Ein tolles Modell ohne Fachwissen oder ohne verständliche Präsentation bringt wenig. — nur auf der Webseite]*

### Analytics oder Science?

#### Data Analytics

Rohdaten holen, ordnen, auswerten. Antwort auf «Was ist passiert?». Ergebnis: Berichte, Grafiken, Empfehlungen.

#### Data Science

Zusätzlich Modelle und Vorhersagen (Machine Learning). Antwort auf «Was passiert als Nächstes?» und «Was sollen wir tun?».

Die Landkarte aus dem Kurs: Beschreibende Analyse trifft *keine* Entscheidungen, Statistik *wenige*, Machine Learning *viele* (automatisch, im grossen Stil).

*[Diagramm: Je weiter rechts, desto mehr Entscheidungen übernimmt der Rechner selbst. — nur auf der Webseite]*

### Algorithmus und Machine Learning

Klassisch programmierst du die Regeln selbst: `Daten + Regeln → Antwort`. Beim Machine Learning gibst du Daten und die gewünschten Antworten, und der Computer findet die Regeln: `Daten + Antworten → Regeln`. Diese gefundenen Regeln nennt man das **Modell**.

*[Diagramm: Blau: was du lieferst. Grün: was herauskommt. Beim Machine Learning ist das Ergebnis die Regel selbst, genannt Modell. — nur auf der Webseite]*

### Wer macht was?

| Rolle | Kernaufgabe |
| --- | --- |
| **Data Analyst** | Daten abfragen (SQL), bereinigen, zusammenfassen, visualisieren. Modelliert nicht. |
| **Data Scientist** | Statistik, Modelle, fragegetriebene Analyse, Resultate an Management erklären. |
| **Data Engineer** | Datenpipelines (ETL), Speicherung, Infrastruktur, Zuverlässigkeit. |
| **ML Engineer** | Algorithmen anwenden, Modelle sauber in Software bringen. |
| **AI Engineer** | Deep Learning, Chatbots und Agenten, Docker, CI/CD, Cloud. |

### CRISP-DM: der Ablauf jedes Projekts

*[Diagramm: CRISP-DM ist ein Kreislauf. Nach der Bewertung geht man oft zurück zu Daten oder Aufbereitung, bis das Ergebnis passt. — nur auf der Webseite]*

Wichtig: Der Weg ist keine Einbahnstrasse. Nach der Bewertung geht man oft zurück zu Schritt 2 oder 3. Und ein Praxistipp aus dem Kurs: ML-Lösungen sind erfolgreicher, wenn sie sich in bestehende Prozesse einfügen, statt neue Prozesse zu erzwingen.

### Die vier ML-Arten

| Art | Idee | Beispiel |
| --- | --- | --- |
| Supervised | Lernen aus Beispielen *mit* Lösung (Label) | Hauspreis vorhersagen, Spam erkennen |
| Unsupervised | Struktur finden, ohne Lösung | Kunden in Gruppen einteilen |
| Semi-supervised | Erst gruppieren, dann Gruppen benennen und als Trainingsdaten nutzen | Wenig Labels, viele Daten |
| Reinforcement | Ausprobieren, Belohnung erhalten | Spiele, Fahren, Trading |

*[Diagramm: Links kennt das Modell die Farbe jedes Punktes und lernt die Trennlinie. Rechts gibt es keine Farben, der Algorithmus sucht die Gruppen selbst. — nur auf der Webseite]*

> **Merksatz**
> Labels (die richtigen Antworten) sind bei Supervised Learning meist das Schwierigste zu bekommen.

### Die «Funktionalen» Werkzeuge

Das sind die Dinge, die am Anfang verwirren. Eigentlich sind es nur Abkürzungen für Schleifen.

| Werkzeug | Was es tut | Beispiel |
| --- | --- | --- |
| `lambda` | Mini-Funktion ohne Namen | `lambda x: x * 2` |
| `map(f, liste)` | wendet `f` auf jedes Element an | `list(map(lambda x: x*2, [1,2,3]))` → `[2,4,6]` |
| `filter(f, liste)` | behält Elemente, bei denen `f` `True` ergibt | `list(filter(lambda x: x>1, [1,2,3]))` → `[2,3]` |
| `zip(a, b)` | paart Elemente aus zwei Listen | `list(zip([1,2],["a","b"]))` → `[(1,'a'),(2,'b')]` |

*[Diagramm: `map` verändert jedes Element, `filter` wirft aus (das gestrichelte Feld fällt weg), `zip` legt zwei Listen paarweise nebeneinander. — nur auf der Webseite]*

### Comprehensions: die Schleife in einer Zeile

```python
# lang:
quadrate = []
for x in range(5):
    if x % 2 == 0:
        quadrate.append(x * x)

# kurz (List Comprehension):  [Ausdruck  for x in ...  if Bedingung]

# Dictionary Comprehension
```

*[Diagramm: Die Comprehension ist ein Fliessband: Zahlen kommen aus `range(5)`, der Filter lässt nur gerade durch, am Ende wird quadriert. — nur auf der Webseite]*

### merge: Tabellen verbinden

Zwei Tabellen haben eine gemeinsame Spalte (den Schlüssel). Mit `pd.merge(a, b, on="schluessel", how=...)` führst du sie zusammen. Das Argument `how` entscheidet, was mit Schlüsseln passiert, die nur in einer Tabelle vorkommen. Probier es aus:

*[Interaktive Demo: Demo — nur auf der Webseite]*

| how= | Behält | Wann nehmen? |
| --- | --- | --- |
| `inner` | nur Schlüssel aus beiden | Zeile nützt nur mit beiden Infos |
| `left` | alle aus der linken Tabelle | Haupttabelle anreichern (am häufigsten) |
| `right` | alle aus der rechten Tabelle | wie left, nur vertauscht |
| `outer` | alle aus beiden | alles sehen, z. B. beim Debuggen |

```
pd.merge(df1, df2, on="property_id", how="inner", validate="one_to_one")
# indicator=True fügt Spalte _merge hinzu: both / left_only / right_only
```

## Pandas unter der Haube

*Modul 1 · Vertiefung*

NumPy, View und Copy, Arrow

Dieses Kapitel erklärt, *warum* Pandas schnell ist und warum manche Fehlermeldungen auftauchen. Es ist Vertiefung: Wenn du sonst genug zu tun hast, überspringe es zuerst.

### Eine Spalte ist ein Array im Speicher

Eine numerische Spalte ist ein **NumPy-Array**: viele Zahlen vom gleichen Typ, direkt hintereinander im Speicher. Eine Python-Liste dagegen enthält nur Verweise (Pointer) auf einzelne Objekte, die irgendwo verstreut liegen.

#### Python-Liste

Jedes Element ist ein eigenes Objekt (rund 28 Byte pro Zahl). Flexibel, aber langsam und schwer.

#### NumPy-Array

Rohe Zahlen, ein Typ, lückenlos gepackt (z. B. 8 Byte pro `int64`). Schnell und kompakt.

Schnell ist NumPy aus drei Gründen: eine kompilierte C-Schleife statt Python-Schleife, zusammenhängender Speicher mit festem Typ und SIMD (die CPU rechnet mit einem Befehl mehrere Zahlen gleichzeitig). Im Kursbeispiel war `x * 2` rund 70-mal schneller als die Listen-Schleife (Messung auf einem Rechner, deiner liefert andere Zahlen).

> **Merksatz**
> «Vektorisiert» (`x * 2`) ist schnell, `.apply()` und Python-Schleifen sind langsam.

### dtype: der Typ bestimmt die Grösse

```
a = np.array([1, 2, 3, 4, 5])   # int64: 8 Byte pro Wert = 40 Byte
a.astype(np.int8)               # 1 Byte pro Wert = 5 Byte, aber nur -128 bis 127!
```

### Broadcasting

Arrays mit verschiedener Form werden passend «gestreckt», ohne zu kopieren. Regel: Formen von rechts anlegen, jede Dimension muss gleich oder 1 sein.

```
col = np.array([[0], [10], [20]])   # Form (3, 1)
row = np.array([1, 2, 3])           # Form (3,)
col + row   # [[ 1, 2, 3],
            #  [11,12,13],
            #  [21,22,23]]
```

### View oder Copy?

Ein **View** ist ein neues Fenster auf dieselben Daten: ändert man den View, ändert sich das Original. Eine **Copy** hat eigene Daten.

| Code | Ergebnis | Warum |
| --- | --- | --- |
| `a[1:4]` | View | normaler Ausschnitt |
| `a[::2]` | View | Schrittweite ist auch ein Ausschnitt |
| `a[[0, 2, 4]]` | Copy | Indexliste, kein einheitlicher Abstand |
| `a[a > 2]` | Copy | Maske sammelt Werte neu ein |
| `a.reshape(2, 3)` | View | nur andere Form, gleiche Bytes |

In Pandas 3 gilt **Copy-on-Write**: Jedes Ergebnis verhält sich wie eine Kopie. Darum ändert verkettetes Zuweisen wie `df["preis"][0] = 100` das `df` nicht mehr. Schreibe mit einem einzigen `loc`:

```
df.loc[0, "preis"] = 100
```

### Strings und Arrow

Früher waren Textspalten `object`: eine Liste von Pointern auf einzelne Python-Strings (langsam, viel Speicher). Pandas 3 speichert Text mit **Apache Arrow** in drei Blöcken: ein Bitfeld für «fehlt», Positionen, wo jeder Text beginnt, und alle Buchstaben hintereinander. Im Kursbeispiel brauchte das für eine Million Strings 13.5 MB statt 58.3 MB.

**Parquet** ist das passende Dateiformat auf der Festplatte: spaltenweise gespeichert, komprimiert, mit Schema. Man liest nur die Spalten, die man braucht.

Kompilierte Schleife, zusammenhängender Speicher mit festem Typ und SIMD. Es gibt keine Objekte und keine Typprüfung pro Element.

</details>

<details><summary>Ist `a[[0, 2, 4]]` ein View oder eine Copy?</summary>

Eine Copy, weil eine Indexliste keinen einheitlichen Abstand beschreibt. `a[::2]` dagegen ist ein View.

</details>

## Supervised Learning: Regression

*Modul 2 · Supervised Learning, Tag 1*

Gradient Descent bis Hyperparameter

Beim Supervised Learning gibst du dem Modell Beispiele mit bekannter Antwort. Bei der **Regression** ist die Antwort eine Zahl (z. B. Hauspreis). Das Modell soll lernen, daraus für *neue* Fälle gut zu raten. Dabei gilt: Die Qualität der Daten begrenzt die Qualität des Modells (*garbage in, garbage out*), und das gilt für die Merkmale genauso wie für die Labels.

*Voraussetzung: Das Kapitel «Mathe-Grundlagen» erklärt, was eine Ableitung und der Gradient sind. Hier geht es direkt los mit dem Lernen.*

### 1 · Gradient Descent: so «lernt» ein Modell

Stell dir einen Nebel auf einem Berg vor. Du willst ins Tal, siehst aber nur, wie steil der Boden unter deinen Füssen ist. Also gehst du immer ein Stück bergab. Genau das macht der Algorithmus.

```
x_neu = x_alt - lernrate * steigung_bei(x_alt)
```

Die **Steigung** (die Ableitung, siehe Kapitel «Mathe-Grundlagen») sagt dir die Richtung. Die **Lernrate** bestimmt die Schrittgrösse. Zu klein: du kommst kaum vorwärts. Zu gross: du springst über das Tal hinaus. Hier ist die Funktion `y = x²`, das Tal liegt bei 0:

*[Interaktive Demo: Demo — nur auf der Webseite]*

Probiere 0.1, 0.5, 1.0 und 1.05. Bei 1.0 springt er nur hin und her, darüber entfernt er sich immer weiter. Bei echten Problemen gibt es zusätzlich viele Täler (lokale Minima). Eine Strategie ist, die Lernrate mit der Zeit zu verkleinern.

### 2 · Lineare Regression: die beste Gerade

Gesucht ist die Gerade `y = a·x + b`, die möglichst nahe an allen Punkten liegt. Wie nah, misst die **Fehlerfunktion** (Loss). Die Abweichung eines Punktes von der Geraden heisst **Residuum**.

| Mass | Bedeutung |
| --- | --- |
| **MAE** | mittlerer absoluter Fehler: Durchschnitt der Beträge der Residuen |
| **MSE** | mittlerer quadrierter Fehler: bestraft grosse Ausreisser viel stärker (RMSE = Wurzel daraus, wieder in der Einheit von y) |
| **R²** | wie viel Prozent des Fehlers die Regression gegenüber «immer den Mittelwert raten» beseitigt. 0 = nicht besser als Mittelwert, 1 = perfekt |

*[Interaktive Demo: Demo — nur auf der Webseite]*

Gradient Descent ist der Weg, wie ein Computer diese Werte findet: mit zufälligen Koeffizienten starten, Fehler messen, Koeffizienten leicht ändern, wiederholen, bis es nicht mehr besser wird. **Mehrere Merkmale** sind einfach mehr Koeffizienten. Für Kurven nimmt man **polynomiale Regression** (x, x², x³ als Merkmale).

#### Beispiel: 10 Wohnungen, mehrere Merkmale

Der Preis ist das Ziel `y`, die Merkmale sind `qm`, `Zimmer` und `Lage` (1 = schlecht bis 5 = top; als Zahl kodiert). Zeichnen kannst du nur zwei Achsen: Hier ist x = qm, y = Preis, die **Punktgrösse** zeigt die Zimmer und die **Farbstärke** die Lage.

*[Interaktive Demo: Demo — nur auf der Webseite]*

Nimm zwei Merkmale (qm und Zimmer), dann gehört zu jeder Wohnung ein Punkt in einem **3D-Raum**: eine Achse pro Merkmal, die dritte Achse ist der Preis. Das Modell ist keine Gerade mehr, sondern eine **Ebene**.

#### Fehlende Werte

- **Löschen**: einfach, aber du verlierst Daten.
- **Auffüllen (Imputation)**: Mittelwert/Median (schnell, ignoriert Zusammenhänge), Modus oder Konstante (auch für Kategorien, kann verzerren), k-nächste-Nachbarn (genauer, teuer), Regression oder Interpolation.

#### Skalieren

Merkmale haben ganz verschiedene Grössenordnungen (Verbrauch 8 l/100km, Hubraum 3600 cc). Ohne Skalieren dominieren die grossen Zahlen. Darum bringt man alle auf eine vergleichbare Skala.

| Methode | Ergebnis |
| --- | --- |
| Standard Scaling | Mittelwert 0, Standardabweichung 1 (Form der Verteilung bleibt) |
| Min-Max Scaling | Werte zwischen 0 und 1 |
| Mean Normalization | Mittelwert 0, Werte zwischen -1 und 1 |

#### Kategorien in Zahlen

Wochentage nie einfach als 1, 2, 3 codieren: Das würde «Mittwoch ist grösser als Montag» behaupten. Stattdessen **One-Hot-Encoding**: pro Kategorie eine Spalte mit 0 oder 1.

```python

X = pd.DataFrame(StandardScaler().fit_transform(data), columns=data.columns)

data["einkommen_klasse"] = pd.cut(data["MedInc"], bins=[-1, 3, 6, 16],
                                  labels=["low", "medium", "high"])

```

> **Falle (Data Leakage)**
> 
> Im Beispiel der Folien wird zur Vereinfachung *vor* dem Aufteilen skaliert. Sauber ist, den Scaler nur auf den Trainingsdaten zu lernen, denn auch der Scaler wird «trainiert»: Er lernt Mittelwert und Streuung. Fliessen Validierungs- oder Testdaten ein, sickert deren Information ins Modell (Data Leakage). Genau dafür gibt es `Pipeline` (siehe unten). Mehr zu Data Leakage kommt am Kurstag zur Klassifikation.

### 4 · Train, Validation, Test und Cross-Validation

Ein Modell, das auf seinen Trainingsdaten gut ist, kann trotzdem bei neuen Daten versagen. Wir wollen die Leistung auf **ungesehenen** Daten kennen (Generalisierung). Deshalb teilen wir die Daten zufällig auf, z. B. 60/20/20:

- **Training (60 %)** Modell lernt hier.
- **Validation (20 %)** Modelle und Einstellungen vergleichen.
- **Test (20 %)** Zum Schluss, nur ein einziges Mal.

Warum nicht gleich den Test benutzen? Weil du die Validation beim Feintunen oft ansiehst und sie dadurch indirekt mitlernt. Der Test ist die Prüfung, die du erst am Schluss schreibst. Danach trainierst du das beste Modell nochmals mit Training und Validation zusammen (mehr Daten helfen).

**Cross-Validation (k-Fold)**: Die Daten werden in k gleich grosse Teile geteilt. k-mal trainierst du auf k−1 Teilen und prüfst auf dem übrigen, dann bildest du den Durchschnitt. Das nutzt die Daten besser, kostet aber k-mal mehr Rechenzeit. Varianten: Stratified (gleiche Klassenanteile pro Teil) und Leave-One-Out.

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score

pipe = Pipeline([("scaler", StandardScaler()), ("model", LinearRegression())])
scores = cross_val_score(pipe, X, y, cv=10)   # 10 Folds
scores.mean(), scores.std()
```

> **Merksatz**
> Pipeline = Schritte (Skalieren, Modell) werden zusammen behandelt, bei jedem Fold nur mit dessen Trainingsteil gelernt. Das verhindert Leakage und macht den Code sauber.

### 5 · Feature Selection

Mehr Merkmale sind nicht automatisch besser. Mit vielen Dimensionen werden Daten «dünn» und Modelle schwerer zu trainieren (*Curse of Dimensionality*). Überflüssige Merkmale erhöhen das Risiko von Overfitting und machen das Modell schwer erklärbar.

- **Filter**: Merkmale nach Statistik oder Korrelation bewerten (z. B. `X.corr()`).
- **Wrapper**: Mit einem Modell verschiedene Merkmalmengen ausprobieren.
- **Embedded**: Die Auswahl steckt im Modell (Lasso, siehe unten).

```python
import seaborn as sns
sns.heatmap(data.corr(), annot=True, cmap="Reds")
```

> **Falle**
> **Falle**
> 
> Korrelation ist keine Ursache. Der Folienwitz mit Piraten und Erderwärmung zeigt es: Zwei Dinge können zusammen laufen, ohne dass das eine das andere verursacht.

### 6 · Bias und Variance: Under- und Overfitting

- **Underfitting (hoher Bias)**: Das Modell ist zu einfach, es verpasst das Muster. Schlecht auf Training *und* Test.
- **Overfitting (hohe Variance)**: Das Modell lernt auch das Rauschen auswendig. Sehr gut auf Training, schlecht auf Test.

Komplexere Modelle haben weniger Bias, aber mehr Variance. Gesucht ist der Mittelweg. Hier ein Polynom verschiedenen Grades, das Punkte einer Sinuskurve mit Rauschen anpasst:

*[Interaktive Demo: Demo — nur auf der Webseite]*

| Diagnose | Training | Test | Gegenmittel |
| --- | --- | --- | --- |
| Underfitting | schlecht | schlecht (ähnlich) | mehr Merkmale, höherer Grad |
| Overfitting | sehr gut | deutlich schlechter | weniger Merkmale, Regularisierung, mehr Daten |

### 7 · Regularisierung

Regularisierung bremst Overfitting: Zur Fehlerfunktion kommt eine **Strafe für grosse Koeffizienten**. Die Stärke regelt ein Parameter (`alpha` bzw. λ). Grosse Koeffizienten sind das typische Zeichen eines Modells, das dem Rauschen nachrennt.

| Methode | Strafe | Wirkung |
| --- | --- | --- |
| **Ridge** (L2) | Summe der quadrierten Koeffizienten | macht alle Koeffizienten kleiner; gut bei vielen korrelierten Merkmalen |
| **Lasso** (L1) | Summe der Beträge der Koeffizienten | setzt unwichtige Koeffizienten auf **exakt 0** (automatische Feature Selection) |
| **ElasticNet** | L1 und L2 gemischt | Kompromiss, robuster als jedes allein |

> **Merksatz**
> Ridge schrumpft, Lasso streicht. Je grösser alpha, desto stärker die Bremse (bei sehr grossem alpha droht Underfitting).

### 8 · Hyperparameter und Modellwahl

**Parameter** lernt das Modell selbst (die Koeffizienten). **Hyperparameter** stellst du vorher ein (Lernrate, alpha, Polynomgrad). Gute Werte findest du durch Suchen, jeweils mit Cross-Validation:

- **Grid Search**: alle Kombinationen aus deiner Liste durchprobieren (gründlich, aber langsam).
- **Random Search**: zufällige Auswahl der Kombinationen (oft fast so gut, viel schneller).

```python
from sklearn.linear_model import Lasso
from sklearn.model_selection import GridSearchCV

pipe = Pipeline([("scaler", StandardScaler()), ("lasso", Lasso())])
grid = {"lasso__alpha": [0.1, 1, 10], "lasso__max_iter": [100, 1000, 10000]}
search = GridSearchCV(pipe, grid, cv=5, n_jobs=-1)
search.fit(X_trainval, y_trainval)
search.best_params_, search.best_estimator_
```

Der Name vor dem doppelten Unterstrich (`lasso__alpha`) ist der Name des Schritts in der Pipeline.

### Der ganze Ablauf auf einen Blick

1. Daten laden, Zielgrösse `y` von den Merkmalen `X` trennen.
2. Test-Set beiseite legen und nicht mehr anfassen.
3. Pipeline bauen (Skalieren, ggf. Features, Modell).
4. Mit Cross-Validation und Grid/Random Search Modelle und Hyperparameter vergleichen.
5. Bestes Modell auf allen Trainingsdaten neu trainieren.
6. Einmal auf dem Test-Set bewerten (MSE, MAE, R²) und fertig.

## Supervised Learning: Klassifikation

*Modul 2 · Supervised Learning, Tag 2 (Fr 30.10.)*

Logistische Regression, Metriken, Ungleichgewicht, Bäume, Ensembles, Leakage

Gleiche Idee wie am Tag 1 (Beispiele mit bekannter Antwort), aber die Antwort ist jetzt eine **Klasse** statt einer Zahl: Betrug ja/nein, Spam ja/nein, Katze/Hund/Vogel. Alles von Tag 1 gilt weiter: Train/Test, Cross-Validation, Skalieren, Regularisierung, Hyperparameter-Suche.

#### Regression

Vorhersage einer Zahl (Preis, Temperatur). Fehler: MSE, MAE.

#### Klassifikation

Vorhersage einer Klasse (ja/nein, A/B/C). Fehler: Cross-Entropy, Metriken wie Precision und Recall.

*Jedes Regressionsproblem lässt sich in Klassen umbauen, indem man das Ziel in Bereiche («Bins») einteilt, zum Beispiel Blutdruck in normal / erhöht / hoch. Das vereinfacht und macht Ergebnisse leichter erklärbar, kostet aber Information.*

### 1 · Logistische Regression

Trotz des Namens ein Klassifikator. Die Idee: Rechne wie bei der linearen Regression `z = w·x + b`. Dieses z kann beliebig gross oder klein sein. Die **Sigmoid-Funktion** presst es in den Bereich 0 bis 1 und macht daraus eine **Wahrscheinlichkeit**:

```
p = 1 / (1 + e^(-z))      z = w1*x1 + w2*x2 + ... + b
```

Aus der Wahrscheinlichkeit wird eine Klasse über eine **Schwelle** (Standard 0,5): p über der Schwelle heisst «positiv», sonst «negativ». Die Stelle, wo p genau 0,5 ist (z = 0), ist die **Entscheidungsgrenze**.

*[Interaktive Demo: Demo — nur auf der Webseite]*

```python
from sklearn.linear_model import LogisticRegression
clf = LogisticRegression().fit(X_train, y_train)
clf.predict(X_test)          # Klassen (0/1)
clf.predict_proba(X_test)    # Wahrscheinlichkeiten pro Klasse
```

> **Merksatz**
> Logistische Regression = lineare Regression + Sigmoid. Das Modell liefert zuerst eine Wahrscheinlichkeit, die Klasse entsteht erst durch die Schwelle. Die Schwelle ist deine Entscheidung, nicht die des Modells.

### 2 · Wie gut ist ein Klassifikator? Confusion Matrix und Metriken

Jede Vorhersage fällt in eines von vier Fächern. Eselsbrücke: Der zweite Buchstabe ist, was das Modell sagt (positiv/negativ), der erste, ob es stimmt (True/False).

|  | Modell sagt positiv | Modell sagt negativ |
| --- | --- | --- |
| **wirklich positiv** | TP (richtig erkannt) | FN (übersehen) |
| **wirklich negativ** | FP (falscher Alarm) | TN (richtig verneint) |

| Mass | Formel | Frage dahinter |
| --- | --- | --- |
| Accuracy | `(TP+TN) / alle` | Wie oft liege ich insgesamt richtig? |
| Precision | `TP / (TP+FP)` | Wenn ich «positiv» sage: wie oft stimmt das? |
| Recall (Sensitivität) | `TP / (TP+FN)` | Wie viele der echten Positiven finde ich? |
| F1 | `2·TP / (2·TP+FP+FN)` | Ein Mittelwert aus Precision und Recall |

> **Falle**
> **Falle (Accuracy)**
> 
> Bei seltenen Klassen ist Accuracy irreführend. Ein Modell, das immer «kein Betrug» sagt, hat bei 1 Betrug auf 100 Fällen 99 % Accuracy, findet aber nie einen Betrug (Recall 0, F1 0). Schau bei unausgewogenen Daten auf Precision, Recall, F1 und AUC.

#### Die Schwelle verschieben: Precision gegen Recall

Senkst du die Schwelle, sagt das Modell öfter «positiv»: mehr TP, aber auch mehr FP. Recall steigt, Precision sinkt. Es ist ein **Kompromiss**, und welche Seite wichtiger ist, hängt vom Problem ab. Bei einer Krebs-Früherkennung willst du kaum etwas übersehen (hoher Recall), bei einer Spam-Sperre willst du keine echte Mail löschen (hohe Precision).

Die **ROC-Kurve** trägt für alle Schwellen die Trefferquote (TPR = Recall) gegen die Rate falscher Alarme (FPR) auf. Ein perfektes Modell liegt oben links, Raten liegt auf der Diagonalen. Die **AUC** ist die Fläche unter der Kurve: 1 ist perfekt, 0,5 ist Zufall.

### 3 · Mehr als zwei Klassen

- **One-vs-All**: Pro Klasse ein Zweiklassen-Modell («diese Klasse gegen alle anderen»). Die Klasse mit der höchsten Wahrscheinlichkeit gewinnt.
- **Softmax**: Die Verallgemeinerung der Sigmoid-Funktion auf mehrere Klassen. Alle Wahrscheinlichkeiten sind positiv und summieren sich zu 1.

Bewertet wird pro Klasse: Du betrachtest jede Klasse der Reihe nach als «positiv» und alle anderen als «negativ» und bekommst so TP, FP, FN, TN und Precision/Recall/F1 je Klasse. Zusammenfassen kannst du das auf zwei Arten:

- **Micro**: Zuerst alle TP, FP, FN über alle Klassen addieren, dann die Masse berechnen. Bei Einzelklassen-Problemen ist micro-Precision = micro-Recall = Accuracy.
- **Macro**: Die Masse pro Klasse berechnen und mitteln (ungewichtet), oder **gewichtet** nach Klassengrösse, damit grosse Klassen mehr zählen.

```python
from sklearn.metrics import classification_report, confusion_matrix
print(confusion_matrix(y_test, pred))
print(classification_report(y_test, pred))   # Precision, Recall, F1 je Klasse + Mittel
```

### 4 · Unausgewogene Klassen

Oft ist das Interessante selten: Betrug, Defekte, Krankheiten. Ein Modell kann dann die Minderheit ignorieren und trotzdem «gut» aussehen. Zwei Dinge sind zu tun: richtig messen (nicht Accuracy) und die Daten oder das Training anpassen.

*[Interaktive Demo: Demo — nur auf der Webseite]*

| Strategie | Idee | Risiko |
| --- | --- | --- |
| Downsampling | Von der Mehrheitsklasse zufällig nur so viele nehmen wie die Minderheit hat | Wirft Information weg |
| Upsampling | Minderheits-Fälle zufällig duplizieren (mit Zurücklegen) | Overfitting auf die Duplikate |
| SMOTE | Neue künstliche Minderheits-Fälle zwischen echten Nachbarn erzeugen: `s = x + (x' − x) · Zufall` | Kann Rauschen erzeugen; braucht `imblearn` |
| Kostengewichtung | Fehler bei der Minderheit zählen mehr (`class_weight="balanced"`) | Wenig, oft die einfachste Wahl |

> **Falle**
> **Falle**
> 
> Resampling (Up/Downsampling, SMOTE) gehört **nur auf die Trainingsdaten**, nie auf das Test-Set. Sonst bewertest du auf künstlichen Daten und bekommst zu schöne Zahlen.

### 5 · Nicht-lineare Klassifikatoren

Logistische Regression zieht eine **gerade** Grenze. Wenn die Klassen nicht so sauber trennbar sind, brauchst du flexiblere Modelle.

#### k-nächste Nachbarn (kNN)

Für einen neuen Punkt suchst du die k nächsten Trainingspunkte und nimmst die Mehrheitsklasse (bei Regression den Durchschnitt). Es gibt kein Training, dafür ist die Vorhersage bei vielen Daten langsam. k steuert Bias und Variance: **k = 1** folgt jedem Punkt (Overfitting), **k = N** sagt für alle die Mehrheitsklasse (Underfitting).

*[Interaktive Demo: Demo — nur auf der Webseite]*

#### Naive Bayes

Berechnet mit dem Satz von Bayes, wie wahrscheinlich eine Klasse bei den beobachteten Merkmalen ist, und tut «naiv» so, als wären die Merkmale voneinander unabhängig. Beispiel aus den Folien: Spielen wir bei Sonne? Von 14 Tagen waren 9 Spieltage und 5 sonnig; an 3 der 9 Spieltage schien die Sonne.

```
P(Ja | sonnig) = P(sonnig | Ja) * P(Ja) / P(sonnig)
               = (3/9) * (9/14) / (5/14) = 0.60
```

#### Support Vector Machine (SVM)

Sucht die Trennlinie mit dem **breitesten Abstand** zu beiden Klassen. Nur die Punkte am Rand (Support Vectors) bestimmen sie. Mit **Kernels** werden die Daten gedanklich in einen höheren Raum gehoben, wo sie sich gerade trennen lassen. Stark bei vielen Merkmalen.

#### Entscheidungsbaum

Eine Kette von Ja/Nein-Fragen. An jedem Knoten wird die Frage gewählt, die die Klassen am besten trennt (Mass: Gini oder Informationsgewinn), bis die Blätter «rein» sind oder die maximale Tiefe erreicht ist. Gut lesbar, aber ein tiefer Baum lernt die Trainingsdaten auswendig. Gegenmittel: `max_depth` begrenzen oder zurückschneiden.

#### Neuronale Netze

Schichten aus künstlichen Neuronen mit gewichteten Verbindungen, die per Backpropagation trainiert werden (Gradient Descent über alle Schichten). Mit sehr vielen Schichten spricht man von Deep Learning, das kommt im nächsten Modul.

### 6 · Ensembles: viele schwache Modelle ergeben ein starkes

Ein einzelnes Modell macht Fehler. Mehrere Modelle, die unterschiedlich irren, gleichen sich im Mittel aus.

#### Bagging (Random Forest)

Viele Bäume, jeder auf einer zufälligen Stichprobe der Daten und einer zufälligen Auswahl an Merkmalen, **parallel** trainiert. Mehrheitsentscheid (oder Mittelwert). Senkt die Varianz.

#### Boosting (Gradient Boosting)

Bäume werden **nacheinander** trainiert, jeder korrigiert die Restfehler des vorherigen. Oft genauer, aber empfindlicher für Rauschen und bei zu hoher Lernrate überangepasst.

```python
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
rf = RandomForestClassifier(n_estimators=200, random_state=0).fit(X_train, y_train)
gb = GradientBoostingClassifier(learning_rate=0.1).fit(X_train, y_train)
```

### 7 · Data Leakage

Leakage heisst: Das Modell sieht beim Training Information, die es bei der echten Vorhersage nicht hätte. Folge: wunderbare Testwerte und ein Modell, das im Einsatz versagt.

- **Target Leakage**: Ein Merkmal enthält die Antwort schon. Beispiel aus der Vorlesung: Ein Kündigungsmodell mit 100 % Treffer, weil Anrufe beim Kundendienst («ich möchte kündigen») erst *nach* der Kündigung entstehen.
- **Zufälliger Schlüssel**: Die Kontonummer sagt etwas voraus, weil sie von der Reihenfolge der Kunden abhängt (z. B. Neukunden haben höhere Nummern).
- **Vorverarbeitung vor dem Split**: Skalieren, Imputieren oder Feature-Auswahl auf allen Daten, bevor getrennt wird. Lösung: `Pipeline`.
- **Zeitreihen**: Zukunft im Training. Immer nach Zeit trennen.

> **Merksatz**
> Misstraue zu guten Zahlen. Frage bei jedem Merkmal: «Hätte ich diesen Wert zum Zeitpunkt der Vorhersage wirklich?» Verdächtig sind ein riesiger Unterschied Training zu Validation und ein einzelnes Merkmal mit enormem Gewicht.

### 8 · Der ganze Ablauf (Titanic-Aufgabe)

Am Nachmittag baut ihr in Gruppen ein Modell, das vorhersagt, wer die Titanic überlebt hat (Spalten: `Pclass`, `Sex`, `Age`, `SibSp`, `Parch`, `Fare`, `Embarked`; Ziel `Survived`). Der Ablauf passt auf jedes Klassifikationsproblem:

1. Daten ansehen (Struktur, Verteilungen, Anteil der Klassen).
2. Aufräumen und Kategorien in Zahlen umwandeln (One-Hot).
3. Neue Merkmale bauen (z. B. Familiengrösse = `SibSp + Parch + 1`).
4. Fehlende Werte füllen (z. B. Alter mit dem Median).
5. Merkmale auswählen.
6. Klassenverhältnis prüfen, bei Bedarf Resampling oder `class_weight`.
7. Train/Test trennen, Cross-Validation auf dem Training.
8. Mit passenden Metriken bewerten (F1, AUC, nicht nur Accuracy).
9. Auf Leakage prüfen.
10. Regularisierung und Hyperparameter-Suche.
11. Bestes Modell **einmal** auf dem Test-Set prüfen.

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score, train_test_split

X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.2, stratify=y, random_state=0)
num = Pipeline([("imp", SimpleImputer(strategy="median")), ("sc", StandardScaler())])
cat = OneHotEncoder(handle_unknown="ignore")
prep = ColumnTransformer([("num", num, ["Age", "Fare"]), ("cat", cat, ["Sex", "Pclass"])])
model = Pipeline([("prep", prep), ("clf", RandomForestClassifier(random_state=0))])
cross_val_score(model, X_tr, y_tr, cv=5, scoring="f1").mean()
```

*`stratify=y` sorgt dafür, dass Train und Test das gleiche Klassenverhältnis haben. Die Pipeline lernt Imputer und Scaler in jedem Fold nur auf dem Trainingsteil und verhindert so Leakage. Die Vorlage des Dozenten liegt im Kursordner unter `02 - SUPERVISED-ML/code/CAS_ML_Supervised_Learning_2.ipynb`.*

## Unsupervised Learning

*Modul 3 · Unsupervised Learning (Sa 31.10.)*

Clustering, Anomalien, PCA und t-SNE

Bisher hatten alle Daten eine Antwort (Preis, Klasse). Beim **Unsupervised Learning** gibt es keine. Das Modell soll selbst Struktur in den Daten finden. Zwei Hauptaufgaben:

#### Clustering

Ähnliche Datenpunkte in Gruppen einteilen. Beispiel: Kundensegmente, Dokumentgruppen.

#### Dimensionsreduktion

Viele Merkmale auf wenige zusammenfassen, zum Beispiel um Daten in 2D zu zeichnen.

Der Tag arbeitet mit einem Lego-Datensatz (von der Spielwarenseite gescrapt). Es gibt kein «richtig», nur nützlich oder nicht. Darum gehört die **Interpretation** zum Ergebnis dazu.

### 1 · Clustering und Anomalien

Ziel: Punkte im gleichen Cluster sind einander ähnlicher als Punkten in anderen Clustern. «Ähnlich» wird über einen **Abstand** gemessen:

- **Euklidische Distanz**: die Luftlinie, `√(Σ (ai − bi)²)`. Für Zahlen.
- **Simple Matching**: zählt, bei wie vielen Kategorien zwei Punkte *nicht* übereinstimmen. Für Kategorien.
- **Dynamic Time Warping**: Abstand zwischen Zeitreihen, die leicht verschoben oder gedehnt sind.

**Anomalieerkennung** ist die Kehrseite: Nach dem Clustering gelten Punkte, die zu keinem Cluster gut passen oder weit von ihrem Zentrum liegen, als auffällig (z. B. ein Betrugsverdacht).

### 2 · k-Means

Du legst die Anzahl Cluster **k** selbst fest. Dann wiederholt der Algorithmus zwei Schritte, bis sich nichts mehr ändert:

1. Jeden Punkt dem **nächsten Zentrum** zuordnen.
2. Jedes Zentrum auf den **Mittelwert** seiner Punkte setzen.

Das Ziel dahinter: die Summe der quadrierten Abstände innerhalb der Cluster (**WSS**, within-cluster sum of squares) möglichst klein machen. Der Start ist zufällig, deshalb kann ein anderer Start ein anderes Ergebnis geben.

*[Interaktive Demo: Demo — nur auf der Webseite]*

> **Falle**
> **Falle (Skalierung)**
> 
> k-Means rechnet mit Abständen. Hat ein Merkmal grosse Zahlen (Preis in Franken) und ein anderes kleine (Bewertung 1 bis 5), entscheidet fast nur das grosse. Darum **vorher skalieren** (`StandardScaler`).

*[Interaktive Demo: Demo — nur auf der Webseite]*

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

X_s = StandardScaler().fit_transform(X)
km = KMeans(n_clusters=3, n_init=10, random_state=0).fit(X_s)
labels = km.labels_            # Cluster pro Zeile
km.cluster_centers_            # Zentren (im skalierten Raum)
km.inertia_                    # WSS
```

### 3 · Wie viele Cluster?

Die Methoden aus dem Supervised Learning (Test-Set, Accuracy) gehen nicht, es gibt ja keine richtige Antwort. Drei übliche Hilfen:

- **Elbow-Methode**: WSS für k = 1, 2, 3 … zeichnen. WSS sinkt immer, aber ab dem «Ellbogen» nur noch wenig. Dort wählst du k.
- **Silhouette**: Pro Punkt: `s = (b − a) / max(a, b)`, mit a = mittlerer Abstand zu den Punkten im *eigenen* Cluster, b = mittlerer Abstand zum *nächsten fremden* Cluster. Nahe 1 passt gut, nahe 0 liegt zwischen zwei Clustern, negativ ist falsch zugeordnet. Du nimmst das k mit dem höchsten Durchschnitt.
- **Gap-Statistik**: Vergleicht deine WSS mit der WSS von Zufallsdaten im gleichen Wertebereich. Gewählt wird das k mit dem grössten Abstand («Gap»).

*[Interaktive Demo: Demo — nur auf der Webseite]*

> **Merksatz**
> Die Methoden liefern Vorschläge, keine Wahrheit. Entscheidend ist am Ende, ob die Cluster fachlich Sinn ergeben und sich beschreiben lassen.

### 4 · Hierarchisches Clustering

Hier musst du k nicht vorher wählen. Jeder Punkt startet als eigenes Cluster, dann werden immer die zwei **ähnlichsten** Cluster verschmolzen, bis alles ein Baum ist, das **Dendrogramm**. Wo du den Baum waagrecht «durchschneidest», bestimmt die Clusterzahl: Je höher der Schnitt, desto weniger Cluster. Die Höhe einer Verbindung zeigt, wie unähnlich die verschmolzenen Gruppen waren.

Wie der Abstand zwischen *Gruppen* gemessen wird, bestimmt die **Linkage**:

| Linkage | Abstand zweier Gruppen | Tendenz |
| --- | --- | --- |
| Single | kleinster Abstand zwischen zwei Punkten | Ketten, lange dünne Cluster |
| Complete | grösster Abstand zwischen zwei Punkten | kompakte, ähnlich grosse Cluster |
| Average | Mittel aller Punktpaare | Kompromiss |
| Ward | so, dass die WSS am wenigsten wächst | kompakt, ähnlich k-Means (häufige Wahl) |

```python
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
Z = linkage(X_s, method="ward")
dendrogram(Z)
labels = fcluster(Z, t=3, criterion="maxclust")   # 3 Cluster
```

*Kleine Entscheide mit grossen Folgen: Standardisieren, Distanzmass, Linkage und Clusterzahl. Ändere jeweils nur eines und vergleiche.*

### 6 · Dimensionsreduktion

Echte Datensätze haben oft Hunderte Merkmale. Das macht Training langsam, Interpretation schwer und führt zum **Fluch der Dimensionen**: In vielen Dimensionen liegen alle Punkte weit auseinander, der Raum ist «leer», und Vorhersagen für neue Punkte werden unzuverlässig.

| Zwei Zufallspunkte im Einheits-… | mittlerer Abstand |
| --- | --- |
| Quadrat (2 Dimensionen) | 0,52 |
| Würfel (3 Dimensionen) | 0,66 |
| Hyperwürfel (1 000 000 Dimensionen) | 408,25 |

Zum Glück liegen die Daten meist nicht gleichmässig im Raum, sondern nahe an einem viel kleineren Unterraum (viele Merkmale sind konstant oder stark korreliert). Zwei Wege:

- **Projektion**: Die Daten auf eine Ebene oder Achse «fallen lassen» (PCA).
- **Manifold Learning**: Die Daten liegen auf einer gebogenen Fläche (wie eine aufgerollte Schweizer Rolle), die man gedanklich glättet (t-SNE).

> **Falle**
> **Falle**
> 
> Jede Reduktion verliert Information. Probiere zuerst mit den Originaldaten, und reduziere nur, wenn es Tempo, Visualisierung oder Rauschen rechtfertigt.

### 7 · PCA (Hauptkomponentenanalyse)

PCA sucht die Richtung, in der die Daten am **stärksten streuen** (die erste Hauptkomponente), dann die nächste Richtung, die darauf senkrecht steht und die restliche Streuung am besten erfasst, usw. Behältst du nur die ersten paar Komponenten, behältst du den grössten Teil der Information in weniger Dimensionen.

*[Interaktive Demo: Demo — nur auf der Webseite]*

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2).fit(X_s)       # vorher skalieren
Z = pca.transform(X_s)                    # neue 2D-Koordinaten
pca.explained_variance_ratio_             # Anteil Varianz je Komponente
```

Mit `explained_variance_ratio_` siehst du, wie viel Information 2 Komponenten behalten. Die Komponenten sind Mischungen der Originalmerkmale und damit schwerer zu deuten.

### 8 · t-SNE

t-SNE ist ein Verfahren fürs **Zeichnen** in 2D. Es versucht, Nachbarschaften zu erhalten: Punkte, die im hohen Raum nah beieinander liegen, sollen auch im 2D-Bild nah liegen. Dazu werden Ähnlichkeiten als Wahrscheinlichkeiten ausgedrückt und die 2D-Punkte per Gradient Descent so verschoben, dass sie dazu passen.

- **Perplexity**: grob die Zahl der Nachbarn, auf die jeder Punkt achtet (typisch 5 bis 50). Der wichtigste Regler, probiere mehrere Werte.
- Weitere Einstellungen: Anzahl Iterationen, Lernrate.

> **Falle**
> **Falle**
> 
> In einem t-SNE-Bild sind **Abstände zwischen Gruppen und Gruppengrössen nicht aussagekräftig**, nur «wer ist mit wem benachbart». Es ist ein Bild zum Erkunden, keine Basis für Modelle, und ein anderer Zufallsstart liefert ein anderes Bild.

```python
from sklearn.manifold import TSNE
Z = TSNE(n_components=2, perplexity=30, random_state=0).fit_transform(X_s)

```

| Nutzen | Vorverarbeitung, Rauschen, 2D-Bild | 2D-Bild von Gruppen |

### Selbsttest

<details><summary>Was ist der Unterschied zwischen Supervised und Unsupervised Learning?</summary>

Supervised lernt aus Beispielen mit bekannter Antwort. Unsupervised hat keine Antwort und sucht selbst Struktur (Gruppen, wichtige Richtungen).

</details>

<details><summary>Welche zwei Schritte wiederholt k-Means?</summary>

Punkte dem nächsten Zentrum zuordnen, dann jedes Zentrum auf den Mittelwert seiner Punkte setzen. Bis sich die Zuordnung nicht mehr ändert.

</details>

<details><summary>Warum musst du vor k-Means skalieren?</summary>

Der Algorithmus rechnet mit Abständen. Ein Merkmal mit grossen Zahlen würde die anderen übertönen.

</details>

<details><summary>Wie findest du die passende Clusterzahl?</summary>

Elbow (WSS gegen k, Knick suchen), Silhouette (höchster Durchschnitt) oder Gap-Statistik. Zusätzlich zählt, ob die Cluster fachlich Sinn ergeben.

</details>

<details><summary>Was ist der Vorteil des hierarchischen Clusterings gegenüber k-Means?</summary>

k muss nicht vorher feststehen: Das Dendrogramm zeigt alle Stufen, und du schneidest auf der gewünschten Höhe. Dafür ist es bei sehr vielen Punkten langsamer.

</details>

<details><summary>Wofür steht die erste Hauptkomponente in PCA?</summary>

Für die Richtung mit der grössten Streuung der Daten. Sie behält am meisten Information, wenn du auf eine Achse reduzierst.

</details>

<details><summary>Was darfst du aus einem t-SNE-Bild nicht ablesen?</summary>

Die Abstände zwischen Gruppen und die Gruppengrössen. Aussagekräftig ist nur, welche Punkte benachbart sind. Zudem hängt das Bild von Perplexity und Zufallsstart ab.

</details>
