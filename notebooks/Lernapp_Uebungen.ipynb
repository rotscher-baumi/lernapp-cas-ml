<details><summary>Lösung zeigen</summary>

```python
def naechstes(p, zentren):
    best, best_d = 0, None
    for j, c in enumerate(zentren):
        d = (p[0] - c[0]) ** 2 + (p[1] - c[1]) ** 2
        if best_d is None or d < best_d:
            best, best_d = j, d
    return best
```
